```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib.colors import Normalize

eos_eps = np.vstack(df['e'].to_numpy())
eos_pres = np.vstack(df['p'].to_numpy())
astro = np.vstack(df['Likelihood_astro'].to_numpy()).flatten()

# First, build segments for each line (Nx26x2)
segments = np.stack([np.stack([eos_eps, eos_pres], axis=2)], axis=1).squeeze(1)  # shape (120000, 26, 2)

# argsort this now
order = np.argsort(astro)

# order the segments
segments_sorted = segments[order]
astro_sorted = astro[order]

# Set up normalization and colormap
norm_astro = (astro_sorted - min(astro_sorted)) / (max(astro_sorted) - min(astro_sorted))
norm = Normalize(vmin=0.0, vmax=1.0)
cmap = plt.cm.jet  # choose your colormap

# Create LineCollection
lc = LineCollection(segments_sorted, cmap=cmap, norm=norm, linewidths=0.5)
lc.set_array(norm_astro)  # associate your values with the collection

# Plot
fig, ax = plt.subplots(figsize=(8,6))
ax.add_collection(lc)
ax.autoscale()  # autoscale the view to the data

# Add a colorbar
cbar = fig.colorbar(lc, ax=ax, label='Astro Likelihood')

ax.set_xlabel(r'$\varepsilon$')
ax.set_ylabel(r'$P$')
ax.set_xscale('log')
ax.set_yscale('log')
plt.show()
```
